```python
import os
import yaml

dataset_path = "/workspace/plant-disease-1"

with open(os.path.join(dataset_path, "data.yaml"), "r") as f:
    data = yaml.safe_load(f)

print("Dataset path:", dataset_path)
print("Classes:", data["names"])
print("Number of classes:", data["nc"])

for split in ["train", "valid", "test"]:
    image_dir = os.path.join(dataset_path, split, "images")
    label_dir = os.path.join(dataset_path, split, "labels")

    images = os.listdir(image_dir)
    labels = os.listdir(label_dir)

    print(f"\n{split}:")
    print("  Images:", len(images))
    print("  Labels:", len(labels))
```

```python
from ultralytics import YOLO

# Load pretrained YOLO11 Nano
model = YOLO("yolo11n.pt")

# Full training
results = model.train(
    data="/workspace/plant-disease-1/data.yaml",
    epochs=100,
    imgsz=640,
    batch=4,
    device=0,
    workers=4,
    project="/workspace/plant-disease-results",
    name="plant_disease_yolo11n",
    exist_ok=True,
    pretrained=True,
    patience=20,
    plots=True,
    save=True
)
```

```python
from ultralytics import YOLO

model = YOLO(
    "/workspace/plant-disease-results/plant_disease_yolo11n/weights/best.pt"
)

test_results = model.val(
    data="/workspace/plant-disease-1/data.yaml",
    split="test",
    imgsz=640,
    batch=4,
    device=0,
    workers=4,
    plots=True,
    project="/workspace/plant-disease-results",
    name="final_test"
)
```

```python
from ultralytics import YOLO

model = YOLO(
    "/workspace/plant-disease-results/plant_disease_yolo11n/weights/best.pt"
)

predictions = model.predict(
    source="/workspace/plant-disease-1/test/images",
    imgsz=640,
    batch=4,
    conf=0.25,
    device=0,
    save=True,
    project="/workspace/plant-disease-results",
    name="test_predictions",
    exist_ok=True
)

print("Prediction images saved.")
```

```python
import shutil
import os

base = "/workspace"

# Create a temporary folder containing dataset + results
project_folder = "/workspace/Plant_Disease_YOLO11_Submission"

os.makedirs(project_folder, exist_ok=True)

shutil.copytree(
    "/workspace/plant-disease-1",
    project_folder + "/dataset",
    dirs_exist_ok=True
)

shutil.copytree(
    "/workspace/plant-disease-results",
    project_folder + "/results",
    dirs_exist_ok=True
)

shutil.make_archive(
    "/workspace/Plant_Disease_YOLO11_Submission",
    "zip",
    project_folder
)

print("Created:")
print("/workspace/Plant_Disease_YOLO11_Submission.zip")
print(
    "Size:",
    round(
        os.path.getsize("/workspace/Plant_Disease_YOLO11_Submission.zip")
        / (1024**3),
        2
    ),
    "GB"
)
```
